You continue with the daily weather of 2023 at the 13 KMI stations of exercise 1, and zoom in on one of them.

```python
import pandas as pd

# The daily KMI observations, as in exercise 1
daily = pd.read_csv("../../data/raw/kmi_daily_2023.csv")
```

## One station

**2.1** Keep only the rows of station `UCCLE`, in Brussels. What was the mean temperature in Uccle in 2023,
and which were its three hottest days?

```python
uccle = daily[daily["name"] == "UCCLE"]
uccle["temp_avg"].mean()
```

```python
uccle.sort_values("temp_max", ascending=False).head(3)
```

12.1 °C on average. The hottest days: 8 July (32.1 °C), and 9 and 10 September, during a late heat wave.

**2.2** Plot the highest temperature of every day in Uccle. Save the plot as a PNG file.

Don't worry too much about the labels on the x-axis: we'll deal with timestamps later. For now, use the row
number (day number) on the x-axis.

Docs: [`plot`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html)

```python
ax = uccle["temp_max"].reset_index(drop=True).plot()
ax.figure.savefig("../../data/uccle.png", dpi=200)
```

**2.3** Make a scatter plot of the highest temperature in Uccle against the minutes of sunshine
(`sun_duration`). Are sunny days warmer?

```python
uccle.plot(x="sun_duration", y="temp_max", kind="scatter")
```

On average yes, but with a lot of spread: with the same amount of sunshine, the highest temperature can
differ by more than 20 °C. The season matters more than the sun on that day.
